```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("FACT_SALES_CLEAN.xlsx")
df.shape
```

```python
df.head()
```

```python
plt.figure(figsize=(6,1))
sns.histplot(df['TOTAL_UNIT_SALES'], kde = True)
plt.title('Distribución Total de Ventas por Unidad')
plt.xlabel('Total de Unidades Vendidas')
plt.ylabel('Frecuencia')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(x='CATEGORY', y='TOTAL_UNIT_SALES', data = df)
plt.title('Distribución Total de Ventas por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Total Unidades Vendidas')
plt.show()
```

```python
weekly_df = df.groupby('WEEK_NUMBER')['TOTAL_UNIT_SALES'].sum().reset_index()

plt.figure(figsize=(12,6))
sns.lineplot(x='WEEK_NUMBER', y='TOTAL_UNIT_SALES', data= weekly_df)
plt.title('Tendencia de Ventas Totales por Semana')
plt.xlabel('Semana del año')
plt.ylabel('Total Unidades Vendidas')
plt.show()
```

```python
segment_df = df.groupby(['WEEK_NUMBER', 'SEGMENT'])['TOTAL_UNIT_SALES'].sum().reset_index()

plt.figure(figsize=(14,7))
sns.lineplot(x='WEEK_NUMBER', y='TOTAL_UNIT_SALES', hue='SEGMENT', data=segment_df)
plt.title('Distribución de ventas por segmentos')
plt.xlabel('Semana del año')
plt.ylabel('Total Unidades Vendidas')
plt.legend(title='Segmento', bbox_to_anchor=(1.05,1),loc='upper left')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='TOTAL_UNIT_SALES', y='TOTAL_VALUE_SALES',data=df, alpha=0.3)
plt.title('Relacion entre unidades y valor de venta')
plt.xlabel('Total Unidades Vendidas')
plt.ylabel('Valor Total de Ventas')
plt.show()
```

```python
region_df = df.groupby('REGION')['TOTAL_UNIT_SALES'].sum().reset_index()

plt.figure(figsize=(12,6))
sns.barplot(y='REGION', x='TOTAL_UNIT_SALES', data=region_df.sort_values('TOTAL_UNIT_SALES', ascending=False))
plt.title('Ventas totales por región')
plt.xlabel('Total Unidades Vendidas')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(x='REGION',y='TOTAL_UNIT_SALES', data=df)

plt.title('Outliers en ventas por región')
plt.xlabel('Región')
plt.ylabel('Unidades vendidas')
plt.xticks(rotation=45)
plt.show()
```
